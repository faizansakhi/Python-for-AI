```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

df = pd.read_csv("Salary_dataset.csv")
df = df.drop(columns=["Unnamed: 0"])
# print(df)

model = LinearRegression()
model.fit(df[["YearsExperience"]], df.Salary)
print(f"Coefficient = {model.coef_[0]}")
print(f"Intercept = {model.intercept_}")

pred = model.predict([[5]])
print(f"Model predicted {round(pred[0], 2)}$ salary")

plt.scatter(df["YearsExperience"], df.Salary, marker = "+", c = "r")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

y_pred = model.predict(df[["YearsExperience"]])
plt.plot(df["YearsExperience"], y_pred, c = "b")
```
