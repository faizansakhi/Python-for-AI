```python
q1 = df["tip"].quantile(0.25)
q3 = df["tip"].quantile(0.75)

iqr = q3 - q1

high = q3 + 1.5 * iqr

no_of_outliers = (df["tip"] > high).sum()

print(f"The upper range of tips is: {high} and the no. of outliers are: {no_of_outliers}")
```
